Question 1: First Recurring CharacterProblem: Given a string s, find and return the first recurring character. If no character repeats, return None.  Example: s = "productscience" $\rightarrow$ Output: "c"

```python
def char(string):
    dict = {}
    for char in string:
        if char not in dict:
            dict[char] = 1
        else:
            if dict[char] > 1:
                return char
            else:
                dict[char] += 1
    return None
```

```python
char('productscience')
```

Question 2: Intersection of Two Unsorted ArraysProblem: Given two integer lists X and Y, write a function to return their intersection (unique common elements) without using Pandas.  Example: X = [1, 5, 9, 0, 5], Y = [3, 0, 2, 9, 9] $\rightarrow$ Output: [0, 9]

```python
X = [1, 5, 9, 0, 5]; Y = [3, 0, 2, 9, 9]
def inter(x, y):
    x = set(x)
    y = set(y)
    return x.intersection(y)
```

```python
inter(X,Y)
```

Level 2: Medium (Two-Pointers, Sliding Windows & Array Logic)Question 3: Maximum Subarray Sum (Kadane’s Algorithm)Problem: Given an integer array, find the sum of the contiguous subarray with the largest sum.  Example: nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4] $\rightarrow$ Output: 6 (from [4, -1, 2, 1])

```python
def max_subarray(nums):
    max_sum = nums[0]
    current_sum = 0
    for num in nums:
        current_sum += num
        if current_sum > max_sum:
            max_sum = current_sum
        if current_sum < 0:
            current_sum = 0
    return max_sum
```

```python
max_subarray([-2, 1, -3, 4, -1, 2, 1, -5, 4])
```
