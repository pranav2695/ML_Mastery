```python
import pandas as pd
import numpy as np

# Sample E-commerce Customer Dataset
data = {
    'Customer_ID': [101, 102, 103, 104, 105, 106, 107, 108],
    'Gender': ['M', 'F', 'f', 'Male', 'Female', 'M', 'F', None],
    'Age': [22, 45, 17, 35, 62, 29, 15, 50],
    'Total_Spend': [150.0, 1200.5, 45.0, 850.0, 3100.0, 400.0, 20.0, 950.0],
    'Payment_Method': ['Credit Card', 'PayPal', 'Cash', 'Credit Card', 'PayPal', 'Cash', 'Cash', 'Credit Card'],
    'Is_Member': [True, False, True, True, False, False, True, False]
}

df = pd.DataFrame(data)
print("--- Initial DataFrame ---")
print(df)
```

```python
def correct_name(row):
    if row.Gender in ['Male', 'M']:
        return 'Male'
    elif row.Gender in ['Female', 'F', 'f']:
        return 'Female'
    else:
        return 'Unknown'
```

```python
df['Gender'] = df.apply(correct_name, axis= 'columns')
df.head()
```

```python
dict = {
    'Male' : ['M', 'Male'], 
    'Female' : ['F', 'f', 'Female']
}
df.map(lambda dict[i] : for i in dict: if x in dict[i])
```

```python
def discount(row):
    if row.Total_Spend > 1000:
        return '20% Discount'
    elif row.Total_Spend > 500:
        return '10% Discount'
    else:
        return 'No Discount'
df['discount_eligibility'] = df.apply(discount, axis='columns')
df.head()
```

```python
df['discount_eligibility_2'] = df.loc[df.Total_Spend]
```

```python
conditions = [
    df['Age'] < 18,
    (df['Age'] >= 18) & (df['Age'] <= 35),
    (df['Age'] > 35) & (df['Age'] <= 55)
]
choices = ['Teen', 'Young Adult', 'Adult']
df['Age_Group'] = np.select(conditions, choices, default='Senior')
```
