```python
def combinationSum(candidates, target: int):
    res = []

    def backtrack(i, path, remaining):
        # 1. BASE CASES
        if remaining == 0:
            res.append(path.copy())  # Found a valid sum! Save a copy of path.
            return
        if remaining < 0:
            return                   # Overshot the target! Stop exploring this path.

        # 2. THE BACKTRACKING LOOP (Choose -> Explore -> Un-choose)
        for j in range(i, len(candidates)):
            candid = candidates[j]
            
            # CHOOSE
            path.append(candid)
            
            # EXPLORE: Fill in the 3 arguments for the recursive call below!
            # Hint: Which index allows reuse of 'candid'? What is the new remaining target?
            backtrack(j, path, target - candid)
            
            # UN-CHOOSE (Backtrack so the next loop iteration starts with a clean slate)
            path.pop()

    # Start recursion at index 0, empty path, and initial target
    backtrack(0, [], target)
    return res
```

```python
def evalRPN(tokens):
    stack = []
    for token in tokens:
        if token not in ['+', '-', '*', '/']:
            stack.append(token)
        else:
            sec_num = stack[-2]; first_num = stack[-1]
            val = eval(str(sec_num + token + first_num))
            stack = stack[:-2]
            stack.append(str(val))
    print(stack[0])
    result = stack[0]
    print(int(result))
    result = int(result)
    return result
```
